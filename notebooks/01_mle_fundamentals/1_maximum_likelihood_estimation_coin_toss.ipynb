```python
# hidden states = which coin is currently being flipped - coin A or coin B
# observations = what we actually see - heads or tails.


'''
We are supposed to identify the actual probability that:
Not whether coin-A or coin-B was used in the 50 trials.
But which coin produced the set of flips (each sequence)


Predefined (incomplete problem scenario) :
N - hidden variables - which coin (A or B)
M - no. of observed symbols - outcomes (Heads or Tails)
pi - initial state distribution. - (Theta_0 for A and B)
A - Transition matrix 
B - Emission matrix
'''
```

```python
import math as mth
# Our dataset: 5 sequences, represented as tuples of (heads, tails)
sequences = [
    (5, 5),
    (9, 1),
    (8, 2),
    (4, 6),
    (7, 3)
]

# Our initial, completely made-up guesses for the coin biases
theta_A = 0.60
theta_B = 0.50


def get_probs(heads,tails,theta_A, theta_B):
    # 1. Calculate raw likelihoods (using ** for exponents)
    likeli_A = (theta_A**heads)*((1-theta_A)**tails)
    likeli_B = (theta_B**heads)*((1-theta_B)**tails)

    prob_A=likeli_A/(likeli_A+likeli_B)    
    prob_B=likeli_B/(likeli_A+likeli_B)
    return prob_A, prob_B

p_A, p_B = get_probs(5,5,theta_A,theta_B)
print(f"prob A:{p_A:.3f}, prob B:{p_B:.3f}")
```

```python
## use the latest expectation to make the better guess(M-step)
# And feed it back to the E-step.

for iteration in range(10):
    # Reset our expected counts to 0 at the start of each loop
    exp_heads_A=0
    exp_tails_A=0
    exp_heads_B=0
    exp_tails_B=0
    
    # E-step: Iterate through all 5 sequences(10-toss x 5-samples)
    for heads, tails in sequences:
        p_A,p_B = get_probs(heads,tails,theta_A,theta_B)
        
        # Add expected flips to Coin A's running total
        exp_heads_A += p_A*heads
        exp_tails_A += p_A*tails
        
        # Add expected flips to Coin B's running total
        exp_heads_B += p_B*heads
        exp_tails_B += p_B*tails
        
    # M-STEP: Update our guesses based on the totals
    # New Theta(A or B) = Exp H/(Exp H + Exp T)
    theta_A = exp_heads_A/(exp_heads_A+exp_tails_A)
    theta_B = exp_heads_B/(exp_heads_B+exp_tails_B)
    
    # Print the result of this iteration to watch it converge!
    print(f"Iteration {iteration + 1}: Theta A = {theta_A:.3f}, Theta B = {theta_B:.3f}")   

    
```

```python
'''
our result shows that:
Theta_A converges to 0.797 ~0.8. This means that 
there is a 79% probability that the outcome is heads given the 
coin tossed is coin-A.

Theta_B converges to 0.520 ~ 0.5. This means that
there is a 52% probability that the outcome is heads given the
coin tossed is coin-B.

We set the number of iterations such that the theta values 
stops changing, which we call as convergence.

'''
```
